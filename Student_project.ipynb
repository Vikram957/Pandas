### Create a new column score_level
"High" if total_score ≥ 90 , "Medium" if 70 ≤ total_score < 90, "Low" if total_score < 70

```python
def score_level(total_score):
  if total_score >=90:
    return 'High'
  elif total_score >=70 and total_score<90:
    return 'Medium'
  else:
    return 'Low'

```

```python
df['score_level'] = df['total_score'].apply(score_level)
```

### Make a new column for study efficiency = total_score ÷ weekly_self_study_hours.

```python
df['study_efficiency'] = (
    df['total_score'] / df['weekly_self_study_hours']
)
```

### Create a new column study_category
"Light" if weekly_self_study_hours < 10,
"Moderate" if 10 ≤ hours ≤ 20,
"Heavy" if hours > 20

```python
def category(c):
  if c <10:
    return 'Light'
  elif c >=10 and c <=20:
    return 'Moderate'
  else:
    return 'Heavy'
```

```python
df['study_category'] = df['weekly_self_study_hours'].apply(category)
```
