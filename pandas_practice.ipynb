DataFrame Modification
1. Create a new column called Grade based on student marks.

```python
def calculate_marks(marks):
  if marks >=90:
    return 'A'
  elif marks >=75:
    return 'B'
  elif marks >=60:
    return 'C'
  elif marks >=45:
    return 'D'
  else:
    return 'Fail'


df['grade'] = df['marks'].apply(calculate_marks)
df

```

Add a column called Bonus to an employee DataFrame.

```python
df['bonus'] = df['salary'] * 0.10
df
```
