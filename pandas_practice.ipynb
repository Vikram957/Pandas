Filtering & Conditional Selection
1. Filter employees whose salary is greater than 30,000.

```python
df[df['salary'] > 30000]
```

Find students who scored more than 80 marks.

```python
df[df['marks'] > 80]
```

Filter employees working in the IT department.

```python
df[df['department']=='it']
```

Find students whose marks are between 60 and 90.

```python
df[(df['marks']>=60) & (df['marks'] <=90)]
```

Filter employees whose age is less than 30 and salary is greater than 25,000.

```python
df[(df['age'] < 30) & (df['salary'] > 25000)]
```

Find customers living in Indore or Bhopal.

```python
df[(df['city'] == 'indore') | (df['city'] == 'bhopal')]
```
