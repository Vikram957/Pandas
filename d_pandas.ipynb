A Series is a one-dimensional labelled array that can store data of different types.

```python
import pandas as pd
import numpy as np
s = pd.Series([10,20,30,40])
print(s)
```

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.Series(data)
print(df)
```

A DataFrame is a two-dimensional labelled data structure containing rows and columns.

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.DataFrame(data)
print(df)
```

Creating Series with custom index

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.Series([10,20,30],index= ["vik",100,3])
print(df)
```

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.Series(data)
print(df)
print(df.dtype)
print(df.ndim)
```

Creating DataFrame with custom index

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.DataFrame(data,['vikram',100])
print(df)
```

Accessing Using Position Index

```python
s = pd.Series([10,20,30,40,50])
print(s[0])
print(s[3])
print(s[4])
```

```python
# Addition Operation
data = pd.Series([1,2,3,4,5])
data +=10
data
```

```python
s = pd.Series([10,20,30,40])
print(s[s>20])
```

```python
# Statistical Opeartions
s = pd.Series([10,20,30,40,50])
print(s)
print(s.sum())
print(s.min())
print(s.max())
print(s.mean())
```
