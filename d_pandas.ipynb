A Series is a one-dimensional labelled array that can store data of different types.

```python
import pandas as pd
s = pd.Series([10,20,30,40])
print(s)
```

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.Series(data)
print(df)
```

A DataFrame is a two-dimensional labelled data structure containing rows and columns.

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.DataFrame(data)
print(df)
```

Creating Series with custom index

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.Series([10,20,30],index= ["vik",100,3])
print(df)
```

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.Series(data)
print(df)
print(df.dtype)
print(df.ndim)
```

Creating DataFrame with custom index

```python
data = {
    "Name" : ["Aman","Riya"],
    "Marks" : [85,99]
}
df = pd.DataFrame(data,['vikram',100])
print(df)
```

```python
data = {
    "Name" : ["Vikram","pulkit","aman","sanjeet","sawan"],
    "Marks": [23,88,78,56,99],
    "Age":[22,21,20,23,20]
}
df = pd.DataFrame(data)
df.to_csv("OUtput.csv")
df.to_excel("excel.xlsx")
df.to_json("Json_File.json")
```
