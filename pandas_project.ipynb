```python
df.isna().sum()
# df['Employee Count'] = df['Employee Count'].fillna(df['Employee Count'].median())

import pandas as pd
import numpy as np
import re

def clean_employee(value):

    if pd.isna(value):
        return np.nan

    value = str(value).lower().strip()

    if value == 'unknown':
        return np.nan

    # Remove commas and employee text
    value = value.replace(',', '')
    value = value.replace('employees', '').replace('employee', '').strip()

    # Remove approximate symbol
    value = value.replace('~', '')

    # Handle values like 11k, 6k
    if value.endswith('k'):
        return float(value[:-1]) * 1000

    # Handle ranges like 1-10, 51-200
    if '-' in value:
        low, high = value.split('-')
        return (float(low) + float(high)) / 2

    # Handle values like 500+, 1000+
    if value.endswith('+'):
        return float(value[:-1])

    # Convert normal numeric values
    return float(value)


df['Employee Count'] = df['Employee Count'].apply(clean_employee)

df['Employee Count'] = df['Employee Count'].fillna(df['Employee Count'].median())
```
